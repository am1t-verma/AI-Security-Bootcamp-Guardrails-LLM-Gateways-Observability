## **Streaming**

```python
from langchain_ollama import ChatOllama

model = ChatOllama(model = "qwen2.5:3b")
```

```python
response = model.invoke("why is rag important?")
print(response.content)
```

```python
# now streaming
response = model.stream("why rag is important?")
response
```

```python
for chunk in response:
    print(chunk.text)
```

```python
response = model.stream("why rag is important?")
response
```

```python
for chunk in response:
    print(chunk.text, end = "", flush = True)
```

## **Batch**

```python
model
```

```python
responses = model.batch([
    "why water is color less?",
    "which model/llm is you are using for response?",
    "why i am lazy always?",
    "give me llm streaming boilerplate code"
])

for i, response in enumerate(responses):
    print(f"Response {i + 1} :\n{response.content}")
```

```python
responses = model.batch([
    "why water is color less?",
    "which model/llm is you are using for response?",
    "why i am lazy always?",
    "give me llm streaming boilerplate code"
],
config = {"max_concurrency" : 2}) # parallel output only 2

for i, response in enumerate(responses):
    print(f"Response {i + 1} :\n{response.content}")
```
